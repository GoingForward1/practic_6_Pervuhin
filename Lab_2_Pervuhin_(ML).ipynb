##**Задание 3.1.** Bootstrap и доверительный интервал

На складе измерили время сборки шести заказов

```python
x = np.array([28, 31, 29, 45, 32, 30])
```

1.Рассчитайте среднее значение исходной выборки

```python
mean_x = np.mean(x)
print(mean_x)
```

2.Реализуйте Bootstrap с количеством итераций.

B = 5000

На каждой итерации:
*   сформируйте выборку такого же размера, как x;
*   используйте выборку с возвращением;
*   рассчитайте её среднее;
*   сохраните результат
.

```python
B = 5000
x_mean_lst = np.zeros(B)
for i in range(B):
  iter = np.random.choice(x, size = len(x), replace = True)
  x_mean_lst[i] = np.mean(iter)
print(x_mean_lst)
```

Получите центральный 95%-й доверительный интервал для среднего

```python
down = np.quantile(x_mean_lst, 0.025)
up = np.quantile(x_mean_lst, 0.975)
print("Доверительный интервал:", down, "-", up)
```

Вопросы для анализа:

Почему Bootstrap используется именно с выборкой с возвращением?

*Потому что мы не имеем настоящей генеральной совокупности, поэтому мы имитируем ее делая случайную выборку с возвращением. Если бы мы делали без возвращения - мы получали бы каждый раз исходную выборку*

 Как наличие наблюдения 45 влияет на Bootstrap-распределение среднего?

 *45 - выборос. В каких то Bootstrap-выборках он может встретиться ни разу а в каких то множество раз из-за чего стандартное отклонение Bootsrap-распределения среднего будет больше, а значит и сильнее разброс.*

  Почему результаты Bootstrap могут заметно изменяться при использовании очень
маленькой исходной выборки?

*Из-за разницы в величине данных*
